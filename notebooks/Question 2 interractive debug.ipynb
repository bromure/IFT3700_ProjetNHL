This notebook is used to look at all the raw data in an user-friendly way, as visualised on a rink at the right position. Note that the raw data should already be in the "data" folder. If not, it could be downloaded using

````python
import ipywidgets as widgets
from IPython.display import display
````

````python
import Visualisation.dataset as data
import Visualisation.question_2 as q2
````

````python
# Year choice
slider_year = widgets.IntSlider(
    value = 2023,
    min = 2016,
    max = 2023,
    step = 1,
    description = "Year:"
)

# Game choice
slider_game = widgets.IntSlider(
    value = 1,
    min = 1,
    max = data.get_n_game(slider_year.value),
    step = 1,
    description = "Game:"
)

# Regular season vs playoffs
checkbox_playoff = widgets.Checkbox(
    value = False,
    description = "Playoffs"
)

def change_slider_game(change) -> None:
    """
    Change the max value of slider_game to reflect the number of game of the selected year
    """
    # TODO: Playoff
    slider_game.max = data.get_n_game(change['new'])

# Affichage
slider_year.observe(change_slider_game, names="value")
box_year = widgets.HBox([slider_year, checkbox_playoff])
display(box_year, slider_game)
q2.get_info(slider_year.value, checkbox_playoff.value, slider_game.value)
````
